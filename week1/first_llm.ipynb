```python
from openai import OpenAI 
import os 
from dotenv import load_dotenv
```

```python
# Load environment variables 
load_dotenv()
```

```python
# Initialize client 
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"), base_url = os.getenv("BASE_URL"))
```

```python
def chat_with_llm(user_input): 
    """Single-turn conversation with LLM""" 
    try: 
        response = client.chat.completions.create( 
            model="gemini-2.5-flash", 
            messages=[ 
                { "role": "system", "content": "You are a helpful assistant."}, 
                { "role": "user", "content": user_input} 
            ], 
            temperature=0.7, 
            max_tokens=150 
        ) 
        return response.choices[0].message.content 
    except Exception as e: 
        return f"Error: {str(e)}"
    
# Test
if __name__ == "__main__": 
    print("LLM assistant started! Type 'exit' to end the conversation.") 
    while True: 
        user_input = input("\nYou: ") 
        if user_input.lower() == 'exit': 
            break
        response = chat_with_llm(user_input) 
        print(f"Assistant: {response}")
```

```python
def demonstrate_parameters(): 
    """Demonstrate the impact of different parameters on output"""

    prompt = "Write a sentence about artificial intelligence"

    # 1. Temperature parameter (creativity) 
    for temp in [0.0, 0.5, 1.0]: 
        response = client.chat.completions.create( 
            model="gemini-2.5-flash", 
            messages=[{"role": "user", "content": prompt}], 
            temperature=temp, 
            n=3  # Generate 3 replies 
        ) 
        print(f"\nTemperature={temp}:") 
    for choice in response.choices: 
            print(f"  - {choice.message.content}")

    # 2. Max_tokens parameter (length control) 
    response = client.chat.completions.create( 
        model="gemini-2.5-flash", 
        messages=[{"role": "user", "content": "Explain machine learning"}], 
        max_tokens=50  # Limit output length 
    )
    print(f"\nOutput limited to 50 tokens: {response.choices[0].message.content}")

    # 3. System message (role setting) 
    roles = [ 
        "You are a rigorous scientist", 
        "You are a 5-year-old child", 
        "You are Shakespeare" 
    ] 
    for role in roles: 
        response = client.chat.completions.create( 
            model="gemini-2.5-flash", 
            messages=[ 
                { "role": "system", "content": role}, 
                { "role": "user", "content": "What is the sun?"} 
            ] 
        ) 
        print(f"\nRole: {role}") 
        print(f"Answer: {response.choices[0].message.content[:100]}...")
```

```python
def stream_response(prompt): 
    """Demonstrate streaming response""" 
    stream = client.chat.completions.create( 
        model="gemini-2.5-flash", 
        messages=[{"role": "user", "content": prompt}], 
        stream=True 
    ) 
    print("Assistant: ", end="")
    collected_messages = []
    for chunk in stream:
        if chunk.choices[0].delta.content:
            content = chunk.choices[0].delta.content
            collected_messages.append(content)
            print(content, end="", flush=True)
    print()
    # Newline 
    return "".join(collected_messages)
```

```python
# Test streaming output 
stream_response("Introduce quantum computing in 100 words") 
```
